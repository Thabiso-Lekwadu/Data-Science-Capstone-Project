* Read in Datasets

```python
df1 = pd.read_excel("C:/Users/amare/OneDrive/Desktop/Pycharm practice/datasets/1st Batch.xlsx", sheet_name="Monthly")
df2 = pd.read_excel("C:/Users/amare/OneDrive/Desktop/Pycharm practice/datasets/2nd Batch.xlsx", sheet_name="Monthly")
df3 = pd.read_excel("C:/Users/amare/OneDrive/Desktop/Pycharm practice/datasets/3rd Batch.xlsx", sheet_name="Monthly")
```

```python
def preprocess(df):
    df = df[4:]
    df = df.drop(range(5, 8))
    df.columns = df.iloc[0]
    df = df.iloc[1:]
    df = df.rename(columns={'title': 'Date'})
    df = df.set_index("Date")
    df.index = pd.to_datetime(df.index)

    return df
```

```python
df1 = preprocess(df1)
df2 = preprocess(df2)
df3 = preprocess(df3)
```

```python
def impute_missing(df):
        df = df.fillna(0)
        df = df.round(0).astype('Int64')
        return df
```

```python
df1 = impute_missing(df1)
df2 = impute_missing(df2)
df3 = impute_missing(df3)
```

Impute missing values

```python
df3.head()
```

```python
data = df1.merge(df2, left_index=True, right_index=True, how='outer').merge(df3, left_index=True, right_index=True, how='outer')
```

```python
data_transposed = data.T.reset_index().rename(columns={4: 'Region'})
data_transposed.head()
```

```python
id_vars = data_transposed.columns[0]
date_cols= data_transposed.columns[1:]
```

```python
data_long = pd.melt(data_transposed, id_vars=id_vars, value_vars=date_cols, var_name='Date', value_name='Value')
data_long.head()
```

```python
data_long[["Province", "Cluster", "Type of Crime"]] = data_long["Region"].str.split("—", expand= True)
data_long["Province"] = data_long["Province"].str.split(":", expand = True).drop(columns= 0)
data_long["Cluster"] = data_long["Cluster"].str.split(":", expand = True).drop(columns= 0)
data_long.head()
```

```python
data_long["Type of Crime"].unique()
```

```python
data_long["Type of Crime"] = data_long["Type of Crime"].replace(' SRCXR: Robbery: Bank, Street and Cash in transit', 'SRCXR: Bank Robbery (Street and Cash in transit)')
```

```python
data_long["Type of Crime"] = data_long["Type of Crime"].str.split(":", expand = True).drop(columns= 0)
```

```python
data_long["Type of Crime"].unique()
```

```python
data_long["Type of Crime"] = data_long["Type of Crime"].str.replace(r'\d+\s*','', regex=True)
```

```python
data_long.drop(columns = "Region", inplace = True)
```

```python
data_long["Date"] = pd.to_datetime(data_long["Date"])
```

```python
data_long["Year"] = data_long["Date"].dt.year
data_long["Month"] = data_long["Date"].dt.month_name()
data_long["Day"] = data_long["Date"].dt.day_name()
data_long = data_long[["Date", "Year", "Month", "Day", "Province", "Cluster", "Type of Crime", "Value"]]
data_long
```

```python
for col in data_long.select_dtypes(include = "object").columns:
    data_long[col] = data_long[col].str.strip()
```
